```python
#  A Linear regression Model using PyTorch

import torch
import torch.optim as optim
import numpy as np
import torch.nn as nn

# Initialize parameters
true_w = 2
true_b = 1
N = 100 # Start with 100 data points
#  Generate random data and split it 80/20
X = np.random.rand(N, 1) 
y = true_w * X + true_b
split_data = int(0.8 * len(X))
x_train, y_train = X[:split_data], y[:split_data]
x_test, y_test = X[split_data:], y[split_data:]

#  Set GPU device
device = "cuda" if torch.cuda.is_available() else "cpu"

# Convert to tensors
x_train_tensor = torch.from_numpy(x_train).float().to(device)
y_train_tensor = torch.from_numpy(y_train).float().to(device)
x_test_tensor = torch.from_numpy(x_test).float().to(device)
y_train_tensor = torch.from_numpy(y_train).float().to(device)
# Set a learning rate
learn_rate = 0.1
# Initialize parameters b & w randomly
torch.manual_seed(42)
b = torch.randn(1, requires_grad=True, dtype=torch.float, device=device)
w = torch.randn(1, requires_grad=True, dtype=torch.float, device=device)
print(f"Initial b & w:\n {b}, {w}")
# Make prediction - forward pass
yhat = w * x_train_tensor + b
# Compute the error and sebsequntly the loss
error = yhat - y_train_tensor
loss = (error ** 2).mean()
print(f"Initial loss is: \n {loss}")
# Compute gradients for parameters b & w
loss.backward()
# Compute gradient
optimizer = optim.SGD([w, b], lr=learn_rate)

```
